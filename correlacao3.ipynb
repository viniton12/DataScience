```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
data = {
    'cidades':['cidade A', 'cidade B', 'cidade C', 'cidade D', 'cidade E'], 
    'números de biblioteca':[5, 8, 3, 6, 4],
    'livros por pessoa por ano':[7.2, 5.5, 6.3, 8.1, 4.9], 
    'porcentagem de leitores por ano':[80, 65, 70, 85, 60]
}

df = pd.DataFrame(data)
print(df)
file_path = r'C:\Users\vinicius\OneDrive\Documentos\correlacao3pratica.xlsx'
df.to_excel(file_path, index=False)
print(f'arquivo exportado{file_path}')
```

```python
plt.figure(figsize=(10,6))
plt.bar(df['números de biblioteca'], df['porcentagem de leitores por ano'], color='skyblue')
plt.title ('relação entre a taxa de leitores e número de biblioteca em cada cidade')
plt.xlabel('números de biblioteca')
plt.ylabel('porcentagem de leitores por ano')
plt.show()
```

```python
#correlação de pyrson
correlacao = df['números de biblioteca'].corr(df['porcentagem de leitores por ano'])
print(correlacao)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['números de biblioteca'], df['porcentagem de leitores por ano'], color='blue', alpha=0.5)
plt.title('relação do número de leitores em relação ao número de biblioteca')
plt.xlabel('números de biblioteca')
plt.ylabel('porcentagem de leitores por ano')
plt.grid('true')
coeficientes = np.polyfit(df['números de biblioteca'], df['porcentagem de leitores por ano'], 1)
polinomio = np.poly1d(coeficientes)
linha_x = np.linspace(df['números de biblioteca'].min(), df['números de biblioteca'].max(), 100)
linha_y = polinomio(linha_x)
plt.plot(linha_x, linha_y, color='red', label='linha de tendência')
plt.legend()
plt.show()
```
