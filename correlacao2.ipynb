```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

#criando a tabela 
data = {
    'Munícipio': ['Itajai','Blumenau', 'Rio do Sul', 'Joinville', 'cutitibanos', 'lages', 'Canoinhas', 'Chapecó', 'concódia', 'Florianópolis', 'Criciúma', 'Laguna'], 
    'população (em 1000 hab)':[1001, 193, 42, 304, 42, 152, 55, 105, 68, 219, 129, 42],
    'pop urbana': [94, 181, 39, 292, 32, 126, 36, 77, 25, 186, 116, 33], 
    '(%) de pop. urbana':[93, 94, 94, 96, 76, 83, 66, 73, 37, 85, 90, 78], 
    'taxa de cresc. demográfico': [3.19, 4.60, 2.78, 6.46, 1.99, 1.89, 2.92, 5.32, 2.71, 3.11, 3.11, 1.21], 
    'taxa de mort. infantil':[ 37, 27, 38, 25, 67, 63, 41, 13, 28, 17, 32, 32], 
    'taxa de alfabetização':[85, 90, 85, 87, 75, 78, 81, 75, 84, 87, 85, 77]
}
df = pd.DataFrame(data)
print(df)
```

```python
#fazendo a correlação entre as colunas  de taxa de alfabetização e a taxa de desenvolvimento urbano
correlacao = df['taxa de alfabetização'].corr(df['taxa de cresc. demográfico'])
print(correlacao)
```

```python
#criando o gráfico de dispersão
plt.figure(figsize = (10,6))
plt.scatter(df['taxa de alfabetização'], df['taxa de cresc. demográfico'], color='blue', alpha = 0.5)
plt.title('correlação')
plt.xlabel('taxa de alfabetização')
plt.ylabel('taxa de cresc. demográfico')
plt.grid ('true')
#criando uma linha de tendência
coeficiente = np.polyfit(df['taxa de alfabetização'], df['taxa de cresc. demográfico'], 1)
polinomio = np.poly1d(coeficiente)
linha_x = np.linspace(df['taxa de alfabetização'].min(), df['taxa de alfabetização'].max(), 100)
linha_y = polinomio (linha_x)

#adicionando a linha de tendência
plt.plot(linha_x, linha_y, color='red', label = 'linha de tendência')
plt.legend()
plt.show()
```
