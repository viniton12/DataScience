```python
import math
import pandas as pd
import numpy as np
```

```python
data = {
    "id do produto": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
 'Categoria': ['Eletrônicos', 'Roupas', 'Eletrônicos', 'Alimentos', 'Roupas', 'Alimentos', 'Eletrônicos', 'Roupas', 'Alimentos', 'Eletrônicos'],
 'preço':[1500, 300, 1000, 20, 400, 10, 800, 250, 15, 2000], 
 'Quantidade vendida':[10, 20, 15, 200, 30, 150, 5, 25, 180, 3]
}
df = pd.DataFrame(data)
print(df)
```

```python
#dados estátisticos 
media_vendas = df['Quantidade vendida'].mean()
preco_medio = df['preço'].mean()
moda = df['Quantidade vendida'].mode()
mediana_vendas = df['Quantidade vendida'].median()
desvio_P = df['Quantidade vendida'].std()
print(' a meida foi', media_vendas)
print('o preço medio foi', preco_medio)
print('a moda foi', moda)
print('a mediana foi', mediana_vendas)
print(' o desvio padrão foi', desvio_P)
```

```python
#análises de vendas
#total de vendas para cada produto
produto101 =  1500 * 10
produto102 = 300 *20
produto103 = 100 * 15
produto104 = 20*200
produto105 = 400*30
produto106 = 10*150
produto107 = 800*5
produto108 = 250*25
produto109 = 15*180
produto110 = 2000*3

print(produto101)
print(produto102)
print(produto103)
print(produto104)
print(produto105)
print(produto106)
print(produto107)
print(produto108)
print(produto109)
print(produto110)


```

```python
#vendas por produtos 
df['Total_Vendas'] = df['preço'] * df['Quantidade vendida']
print(df)
```

```python
#vendas por categoria de produtos
total_vendas_categoria = df.groupby('Categoria')['Total_Vendas'].sum()
print(total_vendas_categoria)
```
